```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files
uploaded = files.upload()

df = pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.csv")

print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nAttrition distribution:")
print(df["Attrition"].value_counts())

print("\nNumerical summary:")
display(df.describe())
```

```python
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

print("\nMissing values by column:")
display(df.isnull().sum().sort_values(ascending=False))

print("\nUnique values in each column:")
display(df.nunique().sort_values())
```

```python
attrition_counts = df["Attrition"].value_counts()

print("Attrition counts:")
print(attrition_counts)

attrition_rate = (df["Attrition"].eq("Yes").mean()) * 100

print(f"\nOverall Attrition Rate: {attrition_rate:.2f}%")
```

```python
sns.set_theme(style="whitegrid")

plt.figure(figsize=(7, 5))

ax = sns.countplot(data=df, x="Attrition", hue="Attrition",
                   palette="Set2", legend=False)

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Age", bins=20, kde=True)

plt.title("Employee Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(data=df, x="MonthlyIncome", bins=30, kde=True)

plt.title("Monthly Income Distribution")
plt.xlabel("Monthly Income")
plt.ylabel("Number of Employees")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="Department",
    hue="Attrition"
)

plt.title("Employee Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="OverTime",
    hue="Attrition"
)

plt.title("Relationship Between Overtime and Attrition")
plt.xlabel("Overtime")
plt.ylabel("Number of Employees")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="JobSatisfaction",
    hue="Attrition"
)

plt.title("Job Satisfaction vs Employee Attrition")
plt.xlabel("Job Satisfaction (1–4)")
plt.ylabel("Number of Employees")
plt.tight_layout()
plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(16, 12))

sns.heatmap(
    numeric_df.corr(),
    cmap="coolwarm",
    center=0,
    linewidths=0.3
)

plt.title("Correlation Heatmap of Employee Features")
plt.tight_layout()
plt.show()
```

```python
df_clean = df.copy()

unnecessary_columns = [
    "EmployeeCount",
    "Over18",
    "StandardHours"
]

df_clean.drop(
    columns=unnecessary_columns,
    inplace=True,
    errors="ignore"
)

print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)
```

```python
print("Missing values:", df_clean.isnull().sum().sum())
print("Duplicate rows:", df_clean.duplicated().sum())

print("\nData types:")
display(df_clean.dtypes)
```

```python
df_clean["Attrition_Flag"] = (
    df_clean["Attrition"].map({"No": 0, "Yes": 1})
)

print(df_clean[["Attrition", "Attrition_Flag"]].head())
```

```python
df_clean["Income_Category"] = pd.cut(
    df_clean["MonthlyIncome"],
    bins=[0, 3000, 6000, 10000, float("inf")],
    labels=["Low", "Medium", "High", "Very High"]
)

df_clean["Age_Group"] = pd.cut(
    df_clean["Age"],
    bins=[17, 25, 35, 45, 55, 100],
    labels=["18-25", "26-35", "36-45", "46-55", "56+"]
)

display(df_clean[
    ["Age", "Age_Group", "MonthlyIncome",
     "Income_Category", "Attrition", "Attrition_Flag"]
].head())
```

```python
output_file = "employee_attrition_cleaned.csv"

df_clean.to_csv(output_file, index=False)

print("Export completed:", output_file)
```
